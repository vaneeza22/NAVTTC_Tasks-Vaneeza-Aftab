# Exception handling


## Handle division and indexing errors


```python
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print('Error: Denominator cannot be 0.')
finally:
    print('This is finally block.')

try:
    even_numbers = [2, 4, 6, 8]
    print(even_numbers[5])
except IndexError:
    print('Index Out of Bound.')
```

## Successful try-else flow


```python
try:
    num = int(input('Enter a number: '))
    if num == 0 or num % 2:
        raise ValueError('Use a nonzero even number')
except ValueError as error:
    print(error)
else:
    print(1 / num)
```
