# Pandas DataFrames


## Create and select rows


```python
import pandas as pd

data = {'calories': [420, 380, 390], 'duration': [50, 40, 45]}
df = pd.DataFrame(data, dtype='int64')
print(df)
print(df.loc[0])
print(df.loc[[0, 1]])
```

## Use named row indexes


```python
import pandas as pd

data = {'calories': [420, 380, 390], 'duration': [50, 40, 45]}
df = pd.DataFrame(data, index=['day1', 'day2', 'day3'],
                  dtype='int64')
print(df)
print(df.loc['day2'])
```
