# Pandas: cleaning data


## Missing cells, dates, wrong values and duplicates


```python
import pandas as pd

df = pd.DataFrame({
    'Duration': [60, 450, 60, 60, 45],
    'Date': ['2020/12/01', '2020/12/08', '2020/12/12',
             '2020/12/12', None],
    'Calories': [409.1, 253.3, 250.7, 250.7, None]
})

print('Rows before:', len(df))
print('Missing calories:', df['Calories'].isna().sum())
print('Duplicates:', df.duplicated().sum())

df.fillna({'Calories': 130}, inplace=True)
df['Date'] = pd.to_datetime(df['Date'], format='mixed')
df.dropna(subset=['Date'], inplace=True)
df.loc[1, 'Duration'] = 45
df.drop_duplicates(inplace=True)

print(df.to_string(index=False))
print('Rows after:', len(df))
```
