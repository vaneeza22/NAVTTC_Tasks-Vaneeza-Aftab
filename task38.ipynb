# NumPy: searching arrays


## Locate matching values


```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 4, 4])
print(np.where(arr == 4)[0])

arr = np.array([10, 14, 93, 41, 8, 7])
print(np.where(arr % 2 == 1)[0])
print(np.where(arr % 2 == 0)[0])
```

## Find insertion positions in a sorted array


```python
import numpy as np

arr = np.array([6, 7, 8, 9])
print(np.searchsorted(arr, 7))
print(np.searchsorted(arr, 7, side='right'))

arr = np.array([1, 3, 5, 7])
print(np.searchsorted(arr, [2, 4, 6]))
```
