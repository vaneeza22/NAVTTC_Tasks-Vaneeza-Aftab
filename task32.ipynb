# NumPy: copy and view


## Independent copy


```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
x = arr.copy()
arr[0] = 42
print(arr)
print(x)
print(x.base)
```

## A view shares the original data


```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
x = arr.view()
arr[0] = 42
print(arr)
print(x)
x[0] = 31
print(arr)
print(x.base)
```
