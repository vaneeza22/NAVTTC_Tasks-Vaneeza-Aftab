```python
# Indexing, slicing and membership
model = 'ChatGPT'
print(model[0])
print(model[4])
print(model[-1])
print(model[-4])
print(model[0:4])
print('Chat' in model)
print(len(model))

```

```python
# Concatenation and immutability
model = 'Opus'
version = '5'
print(model + ' ' + version)
try:
    model[0] = 'X'
except TypeError as error:
    print(type(error).__name__)
print(model)
```
