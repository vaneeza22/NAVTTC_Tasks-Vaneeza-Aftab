Python tuples


```python
# Packing, unpacking and slicing
my_tuple = 3, 4.6, 'dog'
a, b, c = my_tuple
print(a, b, c)
letters = ('p', 'r', 'o', 'g', 'r', 'a', 'm', 'i', 'z')
print(letters[1:4])
print(letters[:-7])
print(letters[7:])
print(type(('hello',)).__name__)
```

```python
# Methods and immutability
letters = ('a', 'p', 'p', 'l', 'e')
print(letters.count('p'))
print(letters.index('l'))
print('a' in letters)
try:
    letters[0] = 'b'
except TypeError as error:
    print(type(error).__name__)
print(letters)
```
