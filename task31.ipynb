# NumPy: data types


## Specify and inspect a data type


```python
import numpy as np

arr = np.array([1, 2, 3, 4], dtype='i4')
print(arr)
print(arr.dtype)

arr = np.array([1, 2, 3, 4], dtype='S')
print(arr)
print(arr.dtype)
```

## Convert existing arrays


```python
import numpy as np

arr = np.array([1.1, 2.1, 3.1])
newarr = arr.astype('i4')
print(newarr)
print(newarr.dtype)

arr = np.array([1, 0, 3])
newarr = arr.astype(bool)
print(newarr)
print(newarr.dtype)
```
