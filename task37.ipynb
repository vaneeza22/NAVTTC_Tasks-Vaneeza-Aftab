# NumPy: splitting arrays


## Even and uneven splits


```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])
for part in np.array_split(arr, 3):
    print(part)
print('Four parts:')
for part in np.array_split(arr, 4):
    print(part)
```

## Split matrix columns


```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
for part in np.hsplit(arr, 3):
    print(part)
```
