# NumPy: sorting arrays


## Numbers, strings, booleans and rows


```python
import numpy as np

print(np.sort(np.array([3, 2, 0, 1])))
print(np.sort(np.array(['banana', 'cherry', 'apple'])))
print(np.sort(np.array([True, False, True])))

arr = np.array([[3, 2, 4], [5, 0, 1]])
print(np.sort(arr))
print(arr)
```
