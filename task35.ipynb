# NumPy: array iteration


## Iterate by row and by element


```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
for row in arr:
    print(row)
for row in arr:
    for value in row:
        print(value, end=' ')
print()
```

## nditer and ndenumerate


```python
import numpy as np

arr = np.array([[1, 2], [3, 4]])
for value in np.nditer(arr):
    print(value)

for index, value in np.ndenumerate(arr):
    print(index, value)
```
