# Mean, median and mode


## Calculate all three measures


```python
import numpy as np
from scipy import stats

speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]
print(f'Mean: {np.mean(speed):.6f}')
print(f'Median: {np.median(speed):.1f}')

result = stats.mode(speed, keepdims=False)
print('Mode:', int(result.mode))
print('Frequency:', int(result.count))
```
