# NumPy: random values


## Random integer and float


```python
from numpy import random

random.seed(0)
print(random.randint(100))

random.seed(0)
print(f'{random.rand():.6f}')
```

## Random arrays and a choice


```python
from numpy import random

random.seed(0)
print(random.randint(100, size=5))

random.seed(0)
print(random.randint(100, size=(3, 5)))

random.seed(0)
print(random.choice([3, 5, 7, 9]))
```
