# Logistic regression


## Load and split the Pima diabetes dataset


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics

columns = ['pregnant', 'glucose', 'bp', 'skin', 'insulin',
           'bmi', 'pedigree', 'age', 'label']
pima = pd.read_csv('pima-indians-diabetes.csv',
                   header=None, names=columns)
features = ['pregnant', 'insulin', 'bmi', 'age',
            'glucose', 'bp', 'pedigree']
X = pima[features]
y = pima['label']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=16)
print('Dataset:', pima.shape)
print('Training:', X_train.shape)
print('Testing:', X_test.shape)
```

## Fit and evaluate the classifier


```python
model = LogisticRegression(random_state=16)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

matrix = metrics.confusion_matrix(y_test, y_pred)
print('Confusion matrix:')
print(matrix)
print(f'Accuracy: {metrics.accuracy_score(y_test, y_pred):.4f}')
print(f'Precision: {metrics.precision_score(y_test, y_pred):.4f}')
print(f'Recall: {metrics.recall_score(y_test, y_pred):.4f}')
print(f'F1: {metrics.f1_score(y_test, y_pred):.4f}')
```
