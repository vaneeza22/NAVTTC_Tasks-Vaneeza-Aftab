# Pandas: read CSV files


## Read the tutorial CSV file


```python
import pandas as pd

url = 'https://www.w3schools.com/python/pandas/data.csv'
df = pd.read_csv(url)
print(df.head().to_string())
print('Rows:', len(df))
print('Columns:', df.columns.tolist())
```

## Select columns while reading


```python
import pandas as pd

url = 'https://www.w3schools.com/python/pandas/data.csv'
df = pd.read_csv(url, usecols=['Duration', 'Calories'], nrows=3)
print(df.to_string(index=False))
```
