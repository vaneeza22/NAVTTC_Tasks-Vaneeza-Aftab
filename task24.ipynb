```python
# Duplicates, adding and removing
numbers = {2, 4, 6, 6, 2, 8}
print(sorted(numbers))
numbers.add(10)
numbers.discard(4)
print(sorted(numbers))
print(len(numbers))
print(type(set()).__name__)
print(type({}).__name__)

# Set operations
A = {1, 3, 5}
B = {1, 2, 3}
print(sorted(A | B))
print(sorted(A & B))
print(sorted(A - B))
print(sorted(A ^ B))
```
