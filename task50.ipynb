# Standard deviation and variance


## A tightly grouped dataset


```python
import numpy as np

speed = [86, 87, 88, 86, 87, 85, 86]
print(f'Standard deviation: {np.std(speed):.6f}')
```

## A more widely spread dataset


```python
import numpy as np

speed = [32, 111, 138, 28, 59, 77, 97]
print(f'Standard deviation: {np.std(speed):.6f}')
print(f'Variance: {np.var(speed):.6f}')
```
