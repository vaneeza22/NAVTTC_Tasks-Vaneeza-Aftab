# Pandas Series


## Default and named indexes


```python
import pandas as pd

a = [1, 7, 2]
myvar = pd.Series(a, dtype='int64')
print(myvar)
print(myvar[0])

myvar = pd.Series(a, index=['x', 'y', 'z'], dtype='int64')
print(myvar)
print(myvar['y'])
```

## Create a Series from a dictionary


```python
import pandas as pd

calories = {'day1': 420, 'day2': 380, 'day3': 390}
myvar = pd.Series(calories, index=['day1', 'day2'], dtype='int64')
print(myvar)
```
