Python dictionaries


```python
# Access, insert, update and delete
capitals = {'Germany': 'Berlin', 'Canada': 'Ottawa'}
print(capitals['Germany'])
capitals['Italy'] = 'Naples'
capitals['Italy'] = 'Rome'
print(capitals)
del capitals['Germany']
print(capitals)
print(len(capitals))

# Iterate over keys and values
capitals = {'England': 'London', 'Italy': 'Rome'}
for country in capitals:
    print(country, capitals[country])
print('Italy' in capitals)
print('France' in capitals)
capitals.clear()
print(capitals)
```
