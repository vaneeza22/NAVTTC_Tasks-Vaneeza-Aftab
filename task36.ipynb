# NumPy: joining arrays


## Concatenate and stack


```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.concatenate((a, b)))
print(np.stack((a, b), axis=1))
print(np.hstack((a, b)))
print(np.vstack((a, b)))
print(np.dstack((a, b)))
```

## Join matrices along columns


```python
import numpy as np

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(np.concatenate((a, b), axis=1))
```
